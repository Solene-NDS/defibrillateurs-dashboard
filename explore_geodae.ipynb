# Step 1 - Exploring the Géo'DAE dataset

**Goal:** get a first look at the national defibrillator database (Géo'DAE) downloaded from data.gouv.fr.

We want to know:
- how big the dataset is,
- which columns it contains,
- how many values are missing,
- whether there are duplicated rows.

Source: https://www.data.gouv.fr/datasets/geodae-base-nationale-des-defibrillateurs (Licence Ouverte 2.0)

## 1. Import the library

We only need **pandas** to read and explore the table.

```python
import pandas as pd
```

## 2. Find the separator

A CSV file can use a comma `,` or a semicolon `;` to separate columns.
We read only the first line of the file and count both characters to find the right one.

```python
# Path to the CSV file (relative to the project folder)
FILE_PATH = "data/raw/geodae.csv"

# Read only the first line of the file
with open(FILE_PATH, encoding="utf-8") as f:
    first_line = f.readline()

# Keep the character that appears the most
if first_line.count(";") > first_line.count(","):
    separator = ";"
else:
    separator = ","

separator
```

## 3. Load the dataset

We load the whole file into a table called a **DataFrame**.
`low_memory=False` avoids warnings when a column contains mixed types.

```python
# Load the full file
df = pd.read_csv(FILE_PATH, sep=separator, encoding="utf-8", low_memory=False)

# Show the size: (number of rows, number of columns)
df.shape
```

## 4. Columns and data types

Each column has a type: `object` = text, `float64` / `int64` = numbers.

```python
# Show all rows of long outputs (there may be many columns)
pd.set_option("display.max_rows", 200)

# List every column with its type
df.dtypes
```

## 5. Missing values

We compute the **percentage of empty cells** in each column, from the highest to the lowest.
This will be useful for the *Data Quality* section of the dashboard.

```python
# Percentage of missing values per column
missing_percent = df.isna().mean() * 100

# Round and sort from highest to lowest
missing_percent.round(1).sort_values(ascending=False)
```

## 6. Duplicated rows

We count rows that are exactly identical on every column.

```python
# Number of fully duplicated rows
df.duplicated().sum()
```

## 7. Example rows

We show 3 rows. `.T` flips the table so that each column appears on its own line, which is easier to read.

```python
# Show the first 3 rows, flipped
df.head(3).T
```
